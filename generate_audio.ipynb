```python
#@title Stap 3 · Audio genereren met de Papiamentu-stem (±10-20 min)
import torch, base64, io, unicodedata
import numpy as np
from transformers import VitsModel, AutoTokenizer
from pydub import AudioSegment
from tqdm.auto import tqdm

model = VitsModel.from_pretrained("facebook/mms-tts-pap")
tok = AutoTokenizer.from_pretrained("facebook/mms-tts-pap")
model.eval()
SR = model.config.sampling_rate

def synth(text):
    t = text.replace("\u2026", " ").strip()
    inputs = tok(t, return_tensors="pt")
    if inputs["input_ids"].shape[-1] == 0:  # onbekende tekens? probeer zonder accenten
        t2 = unicodedata.normalize("NFD", t).encode("ascii", "ignore").decode()
        inputs = tok(t2, return_tensors="pt")
        if inputs["input_ids"].shape[-1] == 0:
            return None
    with torch.no_grad():
        wav = model(**inputs).waveform[0].numpy()
    pcm = (np.clip(wav, -1, 1) * 32767).astype(np.int16)
    seg = AudioSegment(pcm.tobytes(), frame_rate=SR, sample_width=2, channels=1)
    buf = io.BytesIO()
    seg.export(buf, format="mp3", bitrate="48k")
    return base64.b64encode(buf.getvalue()).decode()

AUDIO, mislukt = {}, []
for p in tqdm(PHRASES, desc="Genereren"):
    try:
        b64 = synth(p)
        if b64: AUDIO[p] = "data:audio/mp3;base64," + b64
        else: mislukt.append(p)
    except Exception as e:
        mislukt.append(p)

print(f"✅ {len(AUDIO)} audiofragmenten gemaakt")
if mislukt: print("⚠️ Overgeslagen (app gebruikt daar de reservestem):", mislukt)
```

```python
#@title Stap 4 · Bestand opslaan en downloaden
import json as _json
js = "window.PAP_AUDIO=" + _json.dumps(AUDIO, ensure_ascii=False) + ";"
with open("audio-pap.js", "w", encoding="utf-8") as f:
    f.write(js)
import os
print(f"✅ audio-pap.js gemaakt ({os.path.getsize('audio-pap.js')/1e6:.1f} MB)")

# Luister eerst even naar een voorbeeld:
from IPython.display import Audio as _A, display
voorbeeld = "Bon dia! Kon ta bai?"
if voorbeeld in AUDIO:
    print(f"🔊 Voorbeeld: {voorbeeld}")
    display(_A(url=AUDIO[voorbeeld]))

from google.colab import files
files.download("audio-pap.js")
print("⬇️ Je browser downloadt nu audio-pap.js — dat is het bestand voor GitHub!")
```
